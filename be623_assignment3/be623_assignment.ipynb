```python
#Question 1:Print the type of each of these four values: hbb, the length of hbb, the number 56.67, and the value True. Then state in one sentence what determined each type.

hbb = "ATGGTGCATCTGACTCCTGAGGAGAAGTCTGCCGTTACTGCCCTGTGGGGCAAGGTGAAC"

print(type(hbb))
print(type(len(hbb)))
print(type(56.67))
print(type(True))
```

```python
#In this question, i checked the type of each value using python. I found that each value has a different type depending on the kind of data it contains.
#hbb is text, so it is string. The length is a whole nunber,so it is an integer, 56.67 has a decimal, so it is a float. True is a yes/no type value, so it is a Boolean
```

```python
#Question 2 :Print the first codon, second codon, last three bases, and number of complete codons from hbb.

print(hbb[0:3])
print(hbb[3:6])
print(hbb[-3:])
print(len(hbb) // 3)
```

```python
# Here I used slicing to take the first codon, second codon, and last three bases from the sequence. I also used the length of the sequence to find how many complete codons are present.
```

```python
#Question 3 :Print the RNA version of hbb, and then print the RNA version of the last 12 bases. Also state whether you sliced first or replaced T with U first.

print(hbb.replace("T", "U"))
print(hbb[-12:].replace("T", "U"))
```

```python
# Here,I first changed all the T bases into U to get the RNA sequence. For the last 12 bases, I took the slice first and then changed T to U. This gave me the full RNA sequence and the RNA version of the last 12 bases.
```

```python
#Question 4 :Using only .count() and len(), print:

#How many times G appears in hbb
#How many times GC appears in hbb
#How many bases are neither G nor C.

print(hbb.count("G"))
print(hbb.count("GC"))
print(len(hbb) - hbb.count("G") - hbb.count("C"))
```

```python
# Here, I used count() to find the number of G bases and GC patterns in the sequence. I also used len() along with the G and C counts to find the bases that are neither G nor C
```

```python
#Question 5 : Build a list containing NM_000518, P68871, and 1HHO. Print the length of the list. Then print every item, each preceded by its position starting from 1.

accessions = ["NM_000518", "P68871", "1HHO"]

print(len(accessions))

position = 1

for accession in accessions:
    print(position, accession)
    position = position + 1
```

```python
#Here, I created a list with three accession numbers and checked how many items are present. Then I used a loop with a running position to print each accession starting from 1.
```

```python
#Question 6 :Append ENSG00000244734 to the list from Question 5. Then print the new length and the final item without manually typing the position number or index.

accessions.append("ENSG00000244734")

print(len(accessions))
print(accessions[-1])

```

```python
# Here,I added one more accession number to the existing list using append(). Then I checked the updated length and used -1 to get the last item without manually giving its position.
```
